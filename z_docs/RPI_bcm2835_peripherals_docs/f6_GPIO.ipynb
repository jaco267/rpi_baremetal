- Table 6-2: GPIO **alternate function select register 0**
  - 可以注意到 Bits 對應 gpio.c 的 bit start 
    - GPIO 0 對應 Bits 0~2, bit_start = pin * 3 = 0*3=0    
    - GPIO 1 對應 Bits 3~5, bit start = pin * 3 = 1*3=3  
    - GPIO 9 對應 Bits 27~29, bit start= pin * 3 = 9*3=27
  - Description 對應 gpio.h 的 Gpio func 
    - 0 : in , 1: out , 4: Alt0, 5, Alt1, 6:Alt2

|Bits |Field Name|Description          |Type|Reset|
|-----|----------|---------------------|----|----------|
|31-30|---       |Reserved             |R   |0|
|29-27|FSEL9     |Fsel9- Func Select 9 |R/W |0|
|     |          |000=GPIO Pin9 input  |    | |
|     |          |001=GPIO Pin9 output |    | |
|     |          |100=GPIO Pin9 Alt0   |    | |
|     |          |101=GPIO Pin9 Alt1   |    | |
|     |          |110=GPIO Pin9 Alt2   |    | |
|     |          |111=GPIO Pin9 Alt3   |    | |
|     |          |011=GPIO Pin9 Alt4   |    | |
|     |          |010=GPIO Pin9 Alt5   |    | |
|26-24|FSEL8     |Fsel8-func  Select 8 |R/W |0| 
|...  |FSEL7~1   |...                  |R/W |0|
|2-0  |FSEL0     |FSEL0-func select 0  |R/W |0|

- Table 6-5 : GPIO **alternate function select register 3**

- Table 6-7 : GPIO **alternate function select register 5**
