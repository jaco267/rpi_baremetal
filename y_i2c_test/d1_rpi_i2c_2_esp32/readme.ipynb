### esp32 side  

- i2c sda, scl 其實哪一個 io pin 都可以
```
           IO0  ---1  2 ---IO18 (SPI SCLK)
           IO1  ---3  4 ---IO19 (SPI MISO)       
(SPI MOSI) IO2  ---5  6 ---IO5  (I2C SDA 接到 rpi SDA io2)      
           IO10 ---7  8 ---IO6  (I2C SCL 接到 rpi SCL io3)
(SPI CS)   IO4  ---9  10---IO7
           IO5  ---11 12---3v3 
           gnd  ---13 14---gnd  (gnd 接到 rpi gnd)      
           gnd  ---15 16---gnd     
           gnd  ---17 18---gnd          
           3v3  ---19 20---gnd    
           3v3  ---21 22---IO9   
           gnd  ---23 24---RX         
           en   ---25 26---TX       
           gnd  ---27 28---gnd
           5v   ---29 30---3v3 
```
